```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("assets/luffy.jpg")
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

plt.subplot(2, 1, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2GRAY), cmap='gray')
plt.title("Original Image")
plt.axis("off")
plt.subplot(2, 1, 2)
plt.hist(gray.ravel(), bins=256, range=(0, 256))
plt.title("Grayscale Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")
plt.show()
```

```python
# Brightness and Contrast 
img = cv2.imread("assets/luffy.jpg")

bright = cv2.convertScaleAbs(img, alpha=1.5, beta=50)

constrast = cv2.convertScaleAbs(img, alpha=0.5, beta=0)

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off") 

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(bright, cv2.COLOR_BGR2RGB))
plt.title("Bright Image")
plt.axis("off") 

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(constrast, cv2.COLOR_BGR2RGB))
plt.title("Contrast Image")
plt.axis("off") 

plt.show()
```

```python
# Thresholding
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) 
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
plt.title("Grayscale Image")
plt.axis("off") 

plt.subplot(1, 2, 2)
plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis("off") 

plt.show()
print(gray.shape)
print(gray[100, 100])
```

```python
# adaptive thresholding
adaptive = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 2)

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(adaptive, cmap='gray')
plt.title("Adaptive Threshold Image")
plt.axis("off")

plt.show()
```

```python
# adaptive thresholding gaussian
adaptive = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(adaptive, cmap='gray')
plt.title("Adaptive Threshold Image")
plt.axis("off")

plt.show()
```

```python
#Morphological Operations
from matplotlib.pyplot import subplot


kernel = np.ones((5, 5), np.uint8)
erosion = cv2.erode(binary, kernel, iterations=1)
dilation = cv2.dilate(binary, kernel, iterations=1)

opened = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
closed = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

plt.figure(figsize=(10, 10))
plt.imshow(binary, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.figure(figsize=(10, 10))
subplot(2, 2, 1)
plt.imshow(erosion, cmap='gray')

plt.title("Erosion")
plt.axis("off")

subplot(2, 2, 2)
plt.imshow(dilation, cmap='gray')

plt.title("Dilation")
plt.axis("off")

subplot(2, 2, 3)
plt.imshow(opened, cmap='gray')
plt.title("Opened")
plt.axis("off")

subplot(2, 2, 4)
plt.imshow(closed, cmap='gray')
plt.title("Closed")
plt.axis("off")
```

```python
# Edge Detection
blurred = cv2.GaussianBlur(gray, (5, 5), 0)

edges = cv2.Canny(blurred, 50, 150)

# figure to display the image before blurring and after blurring
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")
plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB), cmap='gray')
plt.title("Blurred Image")
plt.axis("off")

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
plt.title("Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title("Edges")
plt.axis("off")
```

```python
# Contours & basic features
img = cv2.imread('assets/shapes.jpeg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY_INV)

contours, hierarchy = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

result = img.copy()

for contour in contours:
    for contour in contours:

        # Area
        area = cv2.contourArea(contour)

        # Ignore small objects/noise
        if area < 500:
            continue

        # Perimeter
        perimeter = cv2.arcLength(contour, True)

        # Approximate contour
        epsilon = 0.02 * perimeter
        approx = cv2.approxPolyDP(
            contour,
            epsilon,
            True
        )

        # Number of vertices
        vertices = len(approx)

        # Identify shape
        if vertices == 3:
            shape = "Triangle"

        elif vertices == 4:
            shape = "Rectangle/Square"

        elif vertices == 5:
            shape = "Pentagon"

        elif vertices == 6:
            shape = "Hexagon"

        else:
            shape = "Circle/Other"

        # Draw contour
        cv2.drawContours(
            result,
            [contour],
            -1,
            (0, 255, 0),
            2
        )

        # Bounding box
        x, y, w, h = cv2.boundingRect(approx)

        # Put shape name
        cv2.putText(
            result,
            shape,
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 0, 255),
            2
        )

        print("----------------")
        print("Shape:", shape)
        print("Area:", area)
        print("Perimeter:", perimeter)
        print("Vertices:", vertices)
        print("Bounding box:", x, y, w, h)


plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(binary, cv2.COLOR_BGR2RGB),cmap="gray")
plt.title("Binary Image")
plt.axis("off")


plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
plt.title("Contours")
plt.axis("off")
plt.show()

```
