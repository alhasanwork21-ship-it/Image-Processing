```python
import cv2
import os

# Video path
video_path = "video/dhaka_traffic.mp4"

# Folder where frames will be saved
output_folder = "dhaka/frames"
os.makedirs(output_folder, exist_ok=True)

# Open video
cap = cv2.VideoCapture(video_path)

frame_number = 0

while True:
    ret, frame = cap.read()

    # Stop when there are no more frames
    if not ret:
        break

    # Save frame
    filename = os.path.join(
        output_folder,
        f"frame_{frame_number:05d}.jpg"
    )

    cv2.imwrite(filename, frame)

    frame_number += 1

# Release video
cap.release()

print(f"Extracted {frame_number} frames.")
```

```python
# Thresholding

def my_thresholding(img):
    img = cv2.imread(img)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) 
    _, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off") 

    plt.subplot(1, 2, 2)
    plt.imshow(binary, cmap='gray')
    plt.title("Binary Image")
    plt.axis("off") 

    plt.show()
    print(gray.shape)
    print(gray[100, 100])
```

```python
def my_adaptive_thresholding(img):
    # adaptive thresholding gaussian
    img = cv2.imread(img)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    adaptive = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(adaptive, cmap='gray')
    plt.title("Adaptive Threshold Image")
    plt.axis("off")

    plt.show()
```

```python
def my_edge_detection(img):
    # Edge Detection
    img = cv2.imread(img)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)|
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    edges = cv2.Canny(blurred, 50, 150)

    # figure to display the image before blurring and after blurring
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off")
    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB), cmap='gray')
    plt.title("Blurred Image")
    plt.axis("off")

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(edges, cmap='gray')
    plt.title("Edges")
    plt.axis("off")
```

```python
import cv2
import os
import glob

# -----------------------------------
# Paths
# -----------------------------------
input_folder = "frames"
output_video = "output/car_detection_video.mp4"

os.makedirs("output", exist_ok=True)

# -----------------------------------
# Get all frames
# -----------------------------------
frame_paths = sorted(
    glob.glob(os.path.join(input_folder, "*.jpg"))
)

print("Number of frames:", len(frame_paths))

if len(frame_paths) == 0:
    raise ValueError("No frames found!")

# -----------------------------------
# Read first frame
# -----------------------------------
first_frame = cv2.imread(frame_paths[0])

height, width = first_frame.shape[:2]

# -----------------------------------
# FPS
# -----------------------------------
fps = 30

# -----------------------------------
# Video writer
# -----------------------------------
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

video_writer = cv2.VideoWriter(
    output_video,
    fourcc,
    fps,
    (width, height)
)

# -----------------------------------
# Background subtractor
# -----------------------------------
background_subtractor = cv2.createBackgroundSubtractorMOG2(
    history=500,
    varThreshold=50,
    detectShadows=True
)

# -----------------------------------
# Morphology kernel
# -----------------------------------
kernel = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (5, 5)
)

# -----------------------------------
# Process every frame
# -----------------------------------
for i, frame_path in enumerate(frame_paths):

    # Read frame
    frame = cv2.imread(frame_path)

    # --------------------------------
    # Background subtraction
    # --------------------------------
    mask = background_subtractor.apply(frame)

    # --------------------------------
    # Remove shadows
    # --------------------------------
    # MOG2:
    # 0   = background
    # 127 = shadow
    # 255 = foreground

    _, mask = cv2.threshold(
        mask,
        200,
        255,
        cv2.THRESH_BINARY
    )

    # --------------------------------
    # Morphological opening
    # Remove small noise
    # --------------------------------
    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_OPEN,
        kernel,
        iterations=1
    )

    # --------------------------------
    # Closing
    # Connect parts of vehicles
    # --------------------------------
    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        kernel,
        iterations=2
    )

    # --------------------------------
    # Find contours
    # --------------------------------
    contours, _ = cv2.findContours(
        mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    # Copy frame for drawing
    result = frame.copy()

    # --------------------------------
    # Detect objects
    # --------------------------------
    for contour in contours:

        # Area
        area = cv2.contourArea(contour)

        # Ignore small objects
        if area < 500:
            continue

        # Bounding box
        x, y, w, h = cv2.boundingRect(contour)

        # Ignore very small objects
        if w < 20 or h < 20:
            continue

        # Aspect ratio
        aspect_ratio = w / h

        # Ignore extremely thin objects
        if aspect_ratio < 0.3 or aspect_ratio > 5:
            continue

        # --------------------------------
        # Draw bounding box
        # --------------------------------
        cv2.rectangle(
            result,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        # --------------------------------
        # Label
        # --------------------------------
        cv2.putText(
            result,
            "Vehicle",
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 0),
            2
        )

    # --------------------------------
    # Write processed frame
    # --------------------------------
    video_writer.write(result)

    print(
        f"\rProcessing: {i + 1}/{len(frame_paths)}",
        end=""
    )

# -----------------------------------
# Release
# -----------------------------------
video_writer.release()

print("\nDone!")
print("Output:", output_video)
```

```python
import cv2
import os
import time
import glob


def process_traffic_video(input_folder, output_video, fps=30):

    # ============================================================
    # 1. READ TRAFFIC VIDEO FRAMES
    # ============================================================

    frame_paths = sorted(
        glob.glob(os.path.join(input_folder, "*.jpg"))
    )

    if len(frame_paths) == 0:
        raise ValueError(f"No frames found in: {input_folder}")

    print("=" * 60)
    print("TRAFFIC VIDEO PROCESSING")
    print("=" * 60)
    print("Input folder :", input_folder)
    print("Frames       :", len(frame_paths))

    # ============================================================
    # Get frame dimensions
    # ============================================================

    first_frame = cv2.imread(frame_paths[0])

    if first_frame is None:
        raise ValueError("Could not read first frame.")

    height, width = first_frame.shape[:2]

    # ============================================================
    # 2. CREATE OUTPUT VIDEO
    # ============================================================

    os.makedirs(
        os.path.dirname(output_video),
        exist_ok=True
    )

    fourcc = cv2.VideoWriter_fourcc(*"mp4v")

    video_writer = cv2.VideoWriter(
        output_video,
        fourcc,
        fps,
        (width, height)
    )

    # ============================================================
    # BACKGROUND SUBTRACTION
    # ============================================================

    background_subtractor = cv2.createBackgroundSubtractorMOG2(
        history=500,
        varThreshold=50,
        detectShadows=True
    )

    # ============================================================
    # MORPHOLOGICAL KERNEL
    # ============================================================

    kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    # ============================================================
    # VARIABLES
    # ============================================================

    total_objects = 0
    processed_frames = 0

    processing_start = time.time()

    # ============================================================
    # 3. PROCESS EVERY FRAME
    # ============================================================

    for i, frame_path in enumerate(frame_paths):

        frame_start = time.time()

        # Read frame
        frame = cv2.imread(frame_path)

        if frame is None:
            continue

        # --------------------------------------------------------
        # Background subtraction
        # --------------------------------------------------------

        foreground = background_subtractor.apply(frame)

        # --------------------------------------------------------
        # 4. REMOVE NOISE
        # --------------------------------------------------------

        # Remove shadows and keep strong foreground
        _, binary = cv2.threshold(
            foreground,
            200,
            255,
            cv2.THRESH_BINARY
        )

        # Remove small white noise
        binary = cv2.morphologyEx(
            binary,
            cv2.MORPH_OPEN,
            kernel,
            iterations=1
        )

        # Connect broken parts of vehicles
        binary = cv2.morphologyEx(
            binary,
            cv2.MORPH_CLOSE,
            kernel,
            iterations=2
        )

        # Optional dilation to make objects more connected
        binary = cv2.dilate(
            binary,
            kernel,
            iterations=1
        )

        # ========================================================
        # 5. FIND OBJECT CONTOURS
        # ========================================================

        contours, _ = cv2.findContours(
            binary,
            cv2.RETR_EXTERNAL,
            cv2.CHAIN_APPROX_SIMPLE
        )

        frame_object_count = 0

        # Copy original frame for drawing
        result = frame.copy()

        # ========================================================
        # 6. DRAW BOUNDING BOXES
        # ========================================================

        for contour in contours:

            # Calculate area
            area = cv2.contourArea(contour)

            # Ignore small objects/noise
            if area < 500:
                continue

            # Bounding rectangle
            x, y, w, h = cv2.boundingRect(contour)

            # Ignore very small objects
            if w < 20 or h < 20:
                continue

            # Aspect ratio
            aspect_ratio = w / float(h)

            # Ignore extremely thin objects
            if aspect_ratio < 0.3 or aspect_ratio > 5:
                continue

            # Object accepted
            frame_object_count += 1

            # Draw bounding box
            cv2.rectangle(
                result,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

            # Object label
            cv2.putText(
                result,
                "Vehicle",
                (x, y - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 0),
                2
            )

        # ========================================================
        # 7. COUNT DETECTED OBJECTS
        # ========================================================

        total_objects += frame_object_count

        # ========================================================
        # 8. DISPLAY FPS
        # ========================================================

        frame_time = time.time() - frame_start

        if frame_time > 0:
            current_fps = 1.0 / frame_time
        else:
            current_fps = 0

        cv2.putText(
            result,
            f"FPS: {current_fps:.2f}",
            (20, 35),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 255, 255),
            2
        )

        # ========================================================
        # DISPLAY OBJECT COUNT
        # ========================================================

        cv2.putText(
            result,
            f"Objects: {frame_object_count}",
            (20, 70),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 255, 255),
            2
        )

        # ========================================================
        # FRAME NUMBER
        # ========================================================

        cv2.putText(
            result,
            f"Frame: {i + 1}/{len(frame_paths)}",
            (20, 105),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (255, 255, 255),
            2
        )

        # ========================================================
        # 9. WRITE PROCESSED FRAME TO VIDEO
        # ========================================================

        video_writer.write(result)

        processed_frames += 1

        # Progress
        print(
            f"\rProcessing: "
            f"{i + 1}/{len(frame_paths)} | "
            f"Objects: {frame_object_count} | "
            f"FPS: {current_fps:.2f}",
            end=""
        )

    # ============================================================
    # FINISH
    # ============================================================

    video_writer.release()

    processing_time = time.time() - processing_start

    # ============================================================
    # 9. DISPLAY PROCESSING TIME
    # ============================================================

    average_fps = (
        processed_frames / processing_time
        if processing_time > 0
        else 0
    )

    print("\n")
    print("=" * 60)
    print("PROCESSING COMPLETE")
    print("=" * 60)

    print(f"Frames processed : {processed_frames}")
    print(f"Processing time  : {processing_time:.2f} seconds")
    print(f"Average FPS      : {average_fps:.2f}")
    print(f"Total detections : {total_objects}")
    print(f"Output video     : {output_video}")

    print("=" * 60)
```

```python
import cv2
import time
import os

def car_detection(input_video, output_video):

    # -----------------------------
    # 1. Open video
    # -----------------------------
    cap = cv2.VideoCapture(input_video)

    if not cap.isOpened():
        print("❌ Cannot open video:", input_video)
        return

    # -----------------------------
    # 2. Get video information
    # -----------------------------
    fps = cap.get(cv2.CAP_PROP_FPS)

    if fps <= 0:
        fps = 30

    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    print("====================================")
    print("Video:", input_video)
    print("FPS:", fps)
    print("Resolution:", width, "x", height)
    print("Total frames:", total_frames)
    print("====================================")

    # -----------------------------
    # 3. Create output directory
    # -----------------------------
    os.makedirs(os.path.dirname(output_video) or ".", exist_ok=True)

    # -----------------------------
    # 4. Video writer
    # -----------------------------
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")

    out = cv2.VideoWriter(
        output_video,
        fourcc,
        fps,
        (width, height)
    )

    if not out.isOpened():
        print("❌ Cannot create output video")
        cap.release()
        return

    # -----------------------------
    # 5. Background subtraction
    # -----------------------------
    background = cv2.createBackgroundSubtractorMOG2(
        history=500,
        varThreshold=50,
        detectShadows=True
    )

    # -----------------------------
    # 6. Morphological kernel
    # -----------------------------
    kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    frame_number = 0
    total_detections = 0

    start_time = time.time()

    # -----------------------------
    # 7. Process every frame
    # -----------------------------
    while True:

        ret, frame = cap.read()

        if not ret:
            break

        frame_number += 1

        # -----------------------------
        # Detect moving objects
        # -----------------------------
        foreground = background.apply(frame)

        # Remove shadows
        _, binary = cv2.threshold(
            foreground,
            200,
            255,
            cv2.THRESH_BINARY
        )

        # -----------------------------
        # Remove noise
        # -----------------------------
        binary = cv2.morphologyEx(
            binary,
            cv2.MORPH_OPEN,
            kernel,
            iterations=1
        )

        # Fill small gaps
        binary = cv2.morphologyEx(
            binary,
            cv2.MORPH_CLOSE,
            kernel,
            iterations=2
        )

        # Make objects more connected
        binary = cv2.dilate(
            binary,
            kernel,
            iterations=1
        )

        # -----------------------------
        # Find contours
        # -----------------------------
        contours, _ = cv2.findContours(
            binary,
            cv2.RETR_EXTERNAL,
            cv2.CHAIN_APPROX_SIMPLE
        )

        frame_object_count = 0

        # -----------------------------
        # Process contours
        # -----------------------------
        for contour in contours:

            area = cv2.contourArea(contour)

            # Remove very small objects/noise
            if area < 500:
                continue

            x, y, w, h = cv2.boundingRect(contour)

            # Ignore very small boxes
            if w < 20 or h < 20:
                continue

            # Aspect ratio
            aspect_ratio = w / float(h)

            if aspect_ratio < 0.2 or aspect_ratio > 5:
                continue

            # Count object
            frame_object_count += 1

            # Draw bounding box
            cv2.rectangle(
                frame,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

            # Label
            cv2.putText(
                frame,
                "Moving Object",
                (x, y - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 0),
                2
            )

        total_detections += frame_object_count

        # -----------------------------
        # Calculate processing FPS
        # -----------------------------
        elapsed = time.time() - start_time

        processing_fps = frame_number / elapsed if elapsed > 0 else 0

        # -----------------------------
        # Display information
        # -----------------------------
        cv2.putText(
            frame,
            f"Objects: {frame_object_count}",
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 0, 255),
            2
        )

        cv2.putText(
            frame,
            f"FPS: {processing_fps:.2f}",
            (20, 80),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 0, 255),
            2
        )

        cv2.putText(
            frame,
            f"Frame: {frame_number}/{total_frames}",
            (20, 120),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (255, 0, 0),
            2
        )

        # -----------------------------
        # Save processed frame
        # -----------------------------
        out.write(frame)

        # -----------------------------
        # Print progress
        # -----------------------------
        if frame_number % 30 == 0:
            print(
                f"Frame {frame_number}/{total_frames} | "
                f"Objects: {frame_object_count} | "
                f"FPS: {processing_fps:.2f}"
            )

    # -----------------------------
    # 8. Finish
    # -----------------------------
    cap.release()
    out.release()

    total_time = time.time() - start_time

    average_fps = (
        frame_number / total_time
        if total_time > 0
        else 0
    )

    print("\n====================================")
    print("PROCESSING FINISHED")
    print("====================================")
    print("Processed frames:", frame_number)
    print("Processing time:", f"{total_time:.2f}", "seconds")
    print("Average FPS:", f"{average_fps:.2f}")
    print("Total detections:", total_detections)
    print("Output video:", output_video)
    print("====================================")
```

```python
import cv2
import numpy as np
import time
import os


def detect_moving_objects(video_path, output_path):

    # ==========================================
    # 1. Open video
    # ==========================================
    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print("❌ Cannot open video:", video_path)
        return

    fps = cap.get(cv2.CAP_PROP_FPS)

    if fps <= 0:
        fps = 30

    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    print("Video opened successfully")
    print("FPS:", fps)
    print("Resolution:", width, "x", height)
    print("Total frames:", total_frames)

    # ==========================================
    # 2. Create output folder
    # ==========================================
    os.makedirs(
        os.path.dirname(output_path) or ".",
        exist_ok=True
    )

    # ==========================================
    # 3. Video writer
    # ==========================================
    fourcc = cv2.VideoWriter_fourcc(*"mp4v")

    out = cv2.VideoWriter(
        output_path,
        fourcc,
        fps,
        (width, height)
    )

    # ==========================================
    # 4. ORB
    # ==========================================
    orb = cv2.ORB_create(
        nfeatures=2000
    )

    # ==========================================
    # 5. BF Matcher
    # ==========================================
    matcher = cv2.BFMatcher(
        cv2.NORM_HAMMING
    )

    # ==========================================
    # 6. Morphology kernel
    # ==========================================
    kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    # ==========================================
    # 7. First frame
    # ==========================================
    ret, previous_frame = cap.read()

    if not ret:
        print("❌ Cannot read first frame")
        cap.release()
        out.release()
        return

    previous_gray = cv2.cvtColor(
        previous_frame,
        cv2.COLOR_BGR2GRAY
    )

    frame_number = 1
    total_detections = 0

    start_time = time.time()

    # ==========================================
    # 8. Process video
    # ==========================================
    while True:

        ret, current_frame = cap.read()

        if not ret:
            break

        frame_number += 1

        current_gray = cv2.cvtColor(
            current_frame,
            cv2.COLOR_BGR2GRAY
        )

        # ======================================
        # Detect ORB features
        # ======================================
        kp1, des1 = orb.detectAndCompute(
            previous_gray,
            None
        )

        kp2, des2 = orb.detectAndCompute(
            current_gray,
            None
        )

        moving_mask = np.zeros(
            (height, width),
            dtype=np.uint8
        )

        # ======================================
        # Check descriptors
        # ======================================
        if (
            des1 is not None
            and des2 is not None
            and len(kp1) >= 10
            and len(kp2) >= 10
        ):

            # ==================================
            # Match features
            # ==================================
            matches = matcher.knnMatch(
                des1,
                des2,
                k=2
            )

            # ==================================
            # Lowe ratio test
            # ==================================
            good_matches = []

            for pair in matches:

                if len(pair) != 2:
                    continue

                m, n = pair

                if m.distance < 0.75 * n.distance:
                    good_matches.append(m)

            # ==================================
            # Need enough matches
            # ==================================
            if len(good_matches) >= 10:

                # ==================================
                # Get matched coordinates
                # ==================================
                pts_previous = np.float32([
                    kp1[m.queryIdx].pt
                    for m in good_matches
                ]).reshape(-1, 1, 2)

                pts_current = np.float32([
                    kp2[m.trainIdx].pt
                    for m in good_matches
                ]).reshape(-1, 1, 2)

                # ==================================
                # Estimate camera motion
                # ==================================
                H, mask = cv2.findHomography(
                    pts_previous,
                    pts_current,
                    cv2.RANSAC,
                    5.0
                )

                # ==================================
                # If homography exists
                # ==================================
                if H is not None:

                    # ==================================
                    # Align previous frame
                    # ==================================
                    aligned_previous = cv2.warpPerspective(
                        previous_gray,
                        H,
                        (width, height)
                    )

                    # ==================================
                    # Compare aligned/current frame
                    # ==================================
                    difference = cv2.absdiff(
                        aligned_previous,
                        current_gray
                    )

                    # ==================================
                    # Blur
                    # ==================================
                    difference = cv2.GaussianBlur(
                        difference,
                        (5, 5),
                        0
                    )

                    # ==================================
                    # Threshold
                    # ==================================
                    _, moving_mask = cv2.threshold(
                        difference,
                        30,
                        255,
                        cv2.THRESH_BINARY
                    )

                    # ==================================
                    # Remove noise
                    # ==================================
                    moving_mask = cv2.morphologyEx(
                        moving_mask,
                        cv2.MORPH_OPEN,
                        kernel,
                        iterations=1
                    )

                    # ==================================
                    # Close gaps
                    # ==================================
                    moving_mask = cv2.morphologyEx(
                        moving_mask,
                        cv2.MORPH_CLOSE,
                        kernel,
                        iterations=2
                    )

                    # ==================================
                    # Dilate
                    # ==================================
                    moving_mask = cv2.dilate(
                        moving_mask,
                        kernel,
                        iterations=1
                    )

                    # ==================================
                    # Remove border artifacts
                    # ==================================
                    moving_mask[:20, :] = 0
                    moving_mask[-20:, :] = 0
                    moving_mask[:, :20] = 0
                    moving_mask[:, -20:] = 0

        # ======================================
        # Find contours
        # ======================================
        contours, _ = cv2.findContours(
            moving_mask,
            cv2.RETR_EXTERNAL,
            cv2.CHAIN_APPROX_SIMPLE
        )

        object_count = 0

        # ======================================
        # Detect objects
        # ======================================
        for contour in contours:

            area = cv2.contourArea(contour)

            # Remove small noise
            if area < 300:
                continue

            x, y, w, h = cv2.boundingRect(
                contour
            )

            # Remove tiny regions
            if w < 15 or h < 15:
                continue

            object_count += 1

            # ==================================
            # Draw bounding box
            # ==================================
            cv2.rectangle(
                current_frame,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

            cv2.putText(
                current_frame,
                "Object",
                (x, max(y - 8, 20)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (0, 255, 0),
                2
            )

        total_detections += object_count

        # ======================================
        # FPS
        # ======================================
        elapsed = time.time() - start_time

        processing_fps = (
            frame_number / elapsed
            if elapsed > 0
            else 0
        )

        # ======================================
        # Display information
        # ======================================
        cv2.putText(
            current_frame,
            f"Objects: {object_count}",
            (10, 30),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 0, 255),
            2
        )

        cv2.putText(
            current_frame,
            f"FPS: {processing_fps:.2f}",
            (10, 60),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 0, 255),
            2
        )

        cv2.putText(
            current_frame,
            f"Frame: {frame_number}/{total_frames}",
            (10, 90),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (255, 0, 0),
            2
        )

        # ======================================
        # Save frame
        # ======================================
        out.write(current_frame)

        # ======================================
        # Previous = current
        # ======================================
        previous_gray = current_gray.copy()

        # ======================================
        # Progress
        # ======================================
        if frame_number % 25 == 0:

            print(
                f"Frame {frame_number}/{total_frames} | "
                f"Objects: {object_count} | "
                f"FPS: {processing_fps:.2f}"
            )

    # ==========================================
    # Finish
    # ==========================================
    cap.release()
    out.release()

    total_time = time.time() - start_time

    average_fps = (
        frame_number / total_time
        if total_time > 0
        else 0
    )

    print("\n===================================")
    print("PROCESSING FINISHED")
    print("===================================")
    print("Frames processed:", frame_number)
    print("Processing time:", f"{total_time:.2f} seconds")
    print("Average FPS:", f"{average_fps:.2f}")
    print("Total detections:", total_detections)
    print("Output:", output_path)
    print("===================================")
```

```python
import cv2
import csv
import time
import os
import numpy as np


def measure_video(video_path, output_csv):
    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print("Could not open:", video_path)
        return

    # -----------------------------
    # Video information
    # -----------------------------
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    video_fps = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    duration = total_frames / video_fps if video_fps > 0 else 0

    print("\n========== VIDEO INFORMATION ==========")
    print("Video:", video_path)
    print("Width:", width)
    print("Height:", height)
    print("FPS:", video_fps)
    print("Frames:", total_frames)
    print("Duration:", round(duration, 2), "seconds")

    # -----------------------------
    # MOG2
    # -----------------------------
    bg_subtractor = cv2.createBackgroundSubtractorMOG2(
        history=500,
        varThreshold=30,
        detectShadows=False
    )

    kernel = np.ones((5, 5), np.uint8)

    frame_number = 0
    object_counts = []

    processing_start = time.perf_counter()

    while True:

        frame_start = time.perf_counter()

        ret, frame = cap.read()

        if not ret:
            break

        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # Background subtraction
        fg_mask = bg_subtractor.apply(gray)

        # Threshold
        _, binary = cv2.threshold(
            fg_mask,
            127,
            255,
            cv2.THRESH_BINARY
        )

        # Morphological cleaning
        binary = cv2.morphologyEx(
            binary,
            cv2.MORPH_OPEN,
            kernel
        )

        binary = cv2.morphologyEx(
            binary,
            cv2.MORPH_CLOSE,
            kernel
        )

        # Find contours
        contours, _ = cv2.findContours(
            binary,
            cv2.RETR_EXTERNAL,
            cv2.CHAIN_APPROX_SIMPLE
        )

        count = 0

        # Count detected regions
        for contour in contours:

            area = cv2.contourArea(contour)

            if area < 300:
                continue

            count += 1

        object_counts.append(count)

        frame_number += 1

        frame_time = time.perf_counter() - frame_start

    total_processing_time = time.perf_counter() - processing_start

    cap.release()

    # -----------------------------
    # Performance measurements
    # -----------------------------

    processing_fps = (
        frame_number / total_processing_time
        if total_processing_time > 0
        else 0
    )

    average_processing_time = (
        total_processing_time / frame_number
        if frame_number > 0
        else 0
    )

    average_objects = (
        sum(object_counts) / len(object_counts)
        if object_counts
        else 0
    )

    min_objects = min(object_counts) if object_counts else 0
    max_objects = max(object_counts) if object_counts else 0

    # -----------------------------
    # Print results
    # -----------------------------

    print("\n========== PROCESSING RESULTS ==========")

    print("Frames processed:", frame_number)

    print(
        "Total processing time:",
        round(total_processing_time, 3),
        "seconds"
    )

    print(
        "Processing FPS:",
        round(processing_fps, 2)
    )

    print(
        "Average processing time/frame:",
        round(average_processing_time * 1000, 2),
        "ms"
    )

    print(
        "Average detected objects:",
        round(average_objects, 2)
    )

    print(
        "Minimum detected objects:",
        min_objects
    )

    print(
        "Maximum detected objects:",
        max_objects
    )

    # -----------------------------
    # Save CSV
    # -----------------------------

    os.makedirs(
        os.path.dirname(output_csv),
        exist_ok=True
    )

    with open(
        output_csv,
        "w",
        newline=""
    ) as file:

        writer = csv.writer(file)

        writer.writerow([
            "Frame",
            "Object_Count"
        ])

        for i, count in enumerate(object_counts):

            writer.writerow([
                i + 1,
                count
            ])

    print("\nCSV saved to:", output_csv)


# =====================================================
# Run
# =====================================================

measure_video(
    "video/british_highway_traffic.mp4",
    "results/british_highway_traffic.csv"
)
```
