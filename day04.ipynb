```python
import cv2
import os

# Video path
video_path = "video/british_highway_traffic.mp4"

# Folder where frames will be saved
output_folder = "frames"
os.makedirs(output_folder, exist_ok=True)

# Open video
cap = cv2.VideoCapture(video_path)

frame_number = 0

while True:
    ret, frame = cap.read()

    # Stop when there are no more frames
    if not ret:
        break

    # Save frame
    filename = os.path.join(
        output_folder,
        f"frame_{frame_number:05d}.jpg"
    )

    cv2.imwrite(filename, frame)

    frame_number += 1

# Release video
cap.release()

print(f"Extracted {frame_number} frames.")
```

```python
# Thresholding

def my_thresholding(img):
    img = cv2.imread(img)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) 
    _, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off") 

    plt.subplot(1, 2, 2)
    plt.imshow(binary, cmap='gray')
    plt.title("Binary Image")
    plt.axis("off") 

    plt.show()
    print(gray.shape)
    print(gray[100, 100])
```

```python
def my_adaptive_thresholding(img):
    # adaptive thresholding gaussian
    img = cv2.imread(img)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    adaptive = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2)

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(adaptive, cmap='gray')
    plt.title("Adaptive Threshold Image")
    plt.axis("off")

    plt.show()
```

```python
def my_edge_detection(img):
    # Edge Detection
    img = cv2.imread(img)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)|
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    edges = cv2.Canny(blurred, 50, 150)

    # figure to display the image before blurring and after blurring
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off")
    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB), cmap='gray')
    plt.title("Blurred Image")
    plt.axis("off")

    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(gray, cv2.COLOR_BGR2RGB), cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(edges, cmap='gray')
    plt.title("Edges")
    plt.axis("off")
```

```python
import cv2
import os
import glob

# -----------------------------
# Paths
# -----------------------------
input_folder = "frames"
erosion_folder = "erosion_frames"
output_video = "output/erosion_video.mp4"

os.makedirs(erosion_folder, exist_ok=True)
os.makedirs("output", exist_ok=True)

# -----------------------------
# Get frames
# -----------------------------
frame_paths = sorted(
    glob.glob(os.path.join(input_folder, "*.jpg"))
)

print("Number of frames:", len(frame_paths))

# -----------------------------
# First frame
# -----------------------------
first_frame = cv2.imread(frame_paths[0])

height, width = first_frame.shape[:2]

# -----------------------------
# FPS
# -----------------------------
fps = 30

# -----------------------------
# Kernel
# -----------------------------
kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (5, 5)
)

# -----------------------------
# Video writer
# -----------------------------
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

video_writer = cv2.VideoWriter(
    output_video,
    fourcc,
    fps,
    (width, height),
    isColor=False
)

# -----------------------------
# Process frames
# -----------------------------
for i, frame_path in enumerate(frame_paths):

    frame = cv2.imread(frame_path)

    # Grayscale
    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    # Binary image
    _, binary = cv2.threshold(
        gray,
        127,
        255,
        cv2.THRESH_BINARY
    )

    # Erosion
    erosion = cv2.erode(
        binary,
        kernel,
        iterations=1
    )

    # Save
    erosion_path = os.path.join(
        erosion_folder,
        f"frame_{i:05d}.jpg"
    )

    cv2.imwrite(
        erosion_path,
        erosion
    )

    # Video
    video_writer.write(erosion)

    print(
        f"\rProcessing: {i + 1}/{len(frame_paths)}",
        end=""
    )

video_writer.release()

print("\nDone!")
print("Erosion frames:", erosion_folder)
print("Video:", output_video)
```

```python
import cv2
import os
import glob

# -----------------------------------
# Paths
# -----------------------------------
input_folder = "frames"
output_video = "output/car_detection_video.mp4"

os.makedirs("output", exist_ok=True)

# -----------------------------------
# Get all frames
# -----------------------------------
frame_paths = sorted(
    glob.glob(os.path.join(input_folder, "*.jpg"))
)

print("Number of frames:", len(frame_paths))

if len(frame_paths) == 0:
    raise ValueError("No frames found!")

# -----------------------------------
# Read first frame
# -----------------------------------
first_frame = cv2.imread(frame_paths[0])

height, width = first_frame.shape[:2]

# -----------------------------------
# FPS
# -----------------------------------
fps = 30

# -----------------------------------
# Video writer
# -----------------------------------
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

video_writer = cv2.VideoWriter(
    output_video,
    fourcc,
    fps,
    (width, height)
)

# -----------------------------------
# Background subtractor
# -----------------------------------
background_subtractor = cv2.createBackgroundSubtractorMOG2(
    history=500,
    varThreshold=50,
    detectShadows=True
)

# -----------------------------------
# Morphology kernel
# -----------------------------------
kernel = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (5, 5)
)

# -----------------------------------
# Process every frame
# -----------------------------------
for i, frame_path in enumerate(frame_paths):

    # Read frame
    frame = cv2.imread(frame_path)

    # --------------------------------
    # Background subtraction
    # --------------------------------
    mask = background_subtractor.apply(frame)

    # --------------------------------
    # Remove shadows
    # --------------------------------
    # MOG2:
    # 0   = background
    # 127 = shadow
    # 255 = foreground

    _, mask = cv2.threshold(
        mask,
        200,
        255,
        cv2.THRESH_BINARY
    )

    # --------------------------------
    # Morphological opening
    # Remove small noise
    # --------------------------------
    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_OPEN,
        kernel,
        iterations=1
    )

    # --------------------------------
    # Closing
    # Connect parts of vehicles
    # --------------------------------
    mask = cv2.morphologyEx(
        mask,
        cv2.MORPH_CLOSE,
        kernel,
        iterations=2
    )

    # --------------------------------
    # Find contours
    # --------------------------------
    contours, _ = cv2.findContours(
        mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    # Copy frame for drawing
    result = frame.copy()

    # --------------------------------
    # Detect objects
    # --------------------------------
    for contour in contours:

        # Area
        area = cv2.contourArea(contour)

        # Ignore small objects
        if area < 500:
            continue

        # Bounding box
        x, y, w, h = cv2.boundingRect(contour)

        # Ignore very small objects
        if w < 20 or h < 20:
            continue

        # Aspect ratio
        aspect_ratio = w / h

        # Ignore extremely thin objects
        if aspect_ratio < 0.3 or aspect_ratio > 5:
            continue

        # --------------------------------
        # Draw bounding box
        # --------------------------------
        cv2.rectangle(
            result,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        # --------------------------------
        # Label
        # --------------------------------
        cv2.putText(
            result,
            "Vehicle",
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0, 255, 0),
            2
        )

    # --------------------------------
    # Write processed frame
    # --------------------------------
    video_writer.write(result)

    print(
        f"\rProcessing: {i + 1}/{len(frame_paths)}",
        end=""
    )

# -----------------------------------
# Release
# -----------------------------------
video_writer.release()

print("\nDone!")
print("Output:", output_video)
```
